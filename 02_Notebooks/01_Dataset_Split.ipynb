```python
import os
import glob
import shutil
import random
from sklearn.model_selection import train_test_split
```

```python
CLEAN_PATH = "/content/drive/MyDrive/CrisisWeave_Project/01_Dataset/CNN_Severity/Clean"

OUTPUT_PATH = "/content/drive/MyDrive/CrisisWeave_Project/01_Dataset/CNN_Severity"

CLASSES = ["Low", "Moderate", "High"]

random.seed(42)
```

```python
for split in ["Train", "Validation", "Test"]:

    for cls in CLASSES:

        os.makedirs(
            os.path.join(OUTPUT_PATH, split, cls),
            exist_ok=True
        )

print("Folders created successfully.")
```

```python
TRAIN_RATIO = 0.70
VAL_RATIO = 0.15
TEST_RATIO = 0.15

for cls in CLASSES:

    source = os.path.join(CLEAN_PATH, cls)

    images = glob.glob(os.path.join(source, "*"))

    train_imgs, temp_imgs = train_test_split(
        images,
        test_size=0.30,
        random_state=42,
        shuffle=True
    )

    val_imgs, test_imgs = train_test_split(
        temp_imgs,
        test_size=0.50,
        random_state=42,
        shuffle=True
    )

    for img in train_imgs:
        shutil.copy2(
            img,
            os.path.join(OUTPUT_PATH, "Train", cls)
        )

    for img in val_imgs:
        shutil.copy2(
            img,
            os.path.join(OUTPUT_PATH, "Validation", cls)
        )

    for img in test_imgs:
        shutil.copy2(
            img,
            os.path.join(OUTPUT_PATH, "Test", cls)
        )
```

```python
print("=" * 60)
print("DATASET SPLIT SUMMARY")
print("=" * 60)

for split in ["Train", "Validation", "Test"]:

    print(f"\n{split}")

    total = 0

    for cls in CLASSES:

        folder = os.path.join(OUTPUT_PATH, split, cls)

        count = len(glob.glob(os.path.join(folder, "*")))

        total += count

        print(f"{cls:<12}: {count}")

    print(f"Total : {total}")
```
