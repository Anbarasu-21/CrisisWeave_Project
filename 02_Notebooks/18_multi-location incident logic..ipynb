STEP 1 — Add incident-distance calculation + grouping

First patch
Group A center
     ↓
A ──15m── B

C is 80m from B
but if C is > radius from Group A center
→ C becomes another incident

```python
from pathlib import Path
import shutil
from datetime import datetime

path = Path(
    "/content/gdrive/MyDrive/CrisisWeave_Project/05_Streamlit/CrisisWeave_Live/live_engine.py"
)

backup = path.with_name(
    f"live_engine.py.backup_INCIDENT_GROUPING_V1_{datetime.now().strftime('%Y%m%d_%H%M%S')}"
)

shutil.copy2(path, backup)

text = path.read_text(encoding="utf-8")

marker = "def sort_observations(observations):"

if "def haversine_distance_m(" in text:
    print("INCIDENT GROUPING V1 already appears to be installed.")
else:
    patch = '''def haversine_distance_m(lat1, lon1, lat2, lon2):
    """
    Calculate geographic distance between two latitude/longitude
    coordinates using the Haversine formula.

    Returns:
        Distance in meters.
    """
    lat1 = safe_float(lat1)
    lon1 = safe_float(lon1)
    lat2 = safe_float(lat2)
    lon2 = safe_float(lon2)

    if None in (lat1, lon1, lat2, lon2):
        return None

    earth_radius_m = 6371000.0

    phi1 = math.radians(lat1)
    phi2 = math.radians(lat2)

    dphi = math.radians(lat2 - lat1)
    dlambda = math.radians(lon2 - lon1)

    a = (
        math.sin(dphi / 2.0) ** 2
        + math.cos(phi1)
        * math.cos(phi2)
        * math.sin(dlambda / 2.0) ** 2
    )

    return 2.0 * earth_radius_m * math.atan2(
        math.sqrt(a),
        math.sqrt(max(0.0, 1.0 - a))
    )


def group_observations_by_incident(
    observations,
    incident_radius_m=100.0
):
    """
    Group observations into independent geographic incidents.

    Important:
    - Uses observation coordinates.
    - Uses a stable incident center for each group.
    - Does NOT use transitive chaining.
    - Different geographic clusters remain separate.
    """

    valid = []

    for observation in observations:
        lat = safe_float(observation.get("latitude"))
        lon = safe_float(observation.get("longitude"))

        if lat is None or lon is None:
            continue

        item = dict(observation)
        item["_group_lat"] = lat
        item["_group_lon"] = lon
        valid.append(item)

    if not valid:
        return []

    valid = sort_observations(valid)

    groups = []

    for observation in valid:

        lat = observation["_group_lat"]
        lon = observation["_group_lon"]

        assigned_group = None
        assigned_distance = None

        for group in groups:

            distance = haversine_distance_m(
                lat,
                lon,
                group["center_lat"],
                group["center_lon"]
            )

            if (
                distance is not None
                and distance <= incident_radius_m
            ):
                assigned_group = group
                assigned_distance = distance
                break

        if assigned_group is None:

            group_id = f"INCIDENT-{len(groups) + 1:03d}"

            groups.append(
                {
                    "incident_id": group_id,
                    "center_lat": lat,
                    "center_lon": lon,
                    "observations": [observation],
                    "observation_count": 1,
                    "max_observation_distance_m": 0.0
                }
            )

        else:

            assigned_group["observations"].append(
                observation
            )

            assigned_group["observation_count"] += 1

            if assigned_distance is not None:
                assigned_group[
                    "max_observation_distance_m"
                ] = max(
                    assigned_group[
                        "max_observation_distance_m"
                    ],
                    assigned_distance
                )

    for group in groups:

        for observation in group["observations"]:

            observation.pop(
                "_group_lat",
                None
            )

            observation.pop(
                "_group_lon",
                None
            )

    return groups


'''

    if marker not in text:
        raise RuntimeError(
            "Could not find insertion point: "
            "def sort_observations(observations):"
        )

    text = text.replace(
        marker,
        patch + marker,
        1
    )

    path.write_text(
        text,
        encoding="utf-8"
    )

    print("=" * 78)
    print("CRISISWEAVE — INCIDENT GROUPING V1")
    print("=" * 78)
    print(f"Backup created: {backup}")
    print("✓ haversine_distance_m() added")
    print("✓ group_observations_by_incident() added")
    print("✓ Existing reconstruction untouched")
    print("✓ Existing GIS logic untouched")
    print("✓ Existing priority logic untouched")
    print("=" * 78)
```

```python
from pathlib import Path
import re

APP_PATH = Path(
    "/content/gdrive/MyDrive/CrisisWeave_Project/"
    "05_Streamlit/CrisisWeave_Live/live_app.py"
)

text = APP_PATH.read_text(encoding="utf-8")

start_marker = "# ============================================================\n# LIVE RECONSTRUCTION"
end_marker = "# ============================================================\n# LIVE RESULT SUMMARY"

start = text.find(start_marker)
end = text.find(end_marker, start)

if start == -1:
    raise RuntimeError("LIVE RECONSTRUCTION block not found.")

if end == -1:
    raise RuntimeError("LIVE RESULT SUMMARY marker not found.")

new_block = '''# ============================================================
# LIVE RECONSTRUCTION
# ============================================================

st.divider()

st.subheader(
    "5. Live Disaster Reconstruction"
)

st.caption(
    "This executes the complete CrisisWeave pipeline "
    "independently for every detected geographic incident."
)

run_analysis = st.button(
    "🚨 RUN LIVE RECONSTRUCTION",
    type="primary",
    use_container_width=True
)

if run_analysis:

    if not observations:

        st.error(
            "Add at least one image or text "
            "observation first."
        )

    else:

        with st.spinner(
            "Grouping incidents and running "
            "CrisisWeave reconstruction..."
        ):

            try:

                # ------------------------------------------------
                # GROUP OBSERVATIONS INTO INCIDENTS
                # ------------------------------------------------

                incident_groups = (
                    cw.group_observations_by_incident(
                        observations,
                        incident_radius_m=100.0
                    )
                )

                if not incident_groups:

                    raise ValueError(
                        "No valid geographic incident groups "
                        "could be created from the observations."
                    )

                analysis_time = (
                    datetime.now(
                        timezone.utc
                    ).isoformat()
                )

                incident_results = {}

                # ------------------------------------------------
                # RECONSTRUCT EACH INCIDENT INDEPENDENTLY
                # ------------------------------------------------

                for group in incident_groups:

                    incident_id = group[
                        "incident_id"
                    ]

                    group_lat = float(
                        group["center_lat"]
                    )

                    group_lon = float(
                        group["center_lon"]
                    )

                    group_observations = group[
                        "observations"
                    ]

                    # --------------------------------------------
                    # LIVE GIS FOR THIS INCIDENT
                    # --------------------------------------------

                    group_gis_context = (
                        cw.fetch_live_gis(
                            group_lat,
                            group_lon,
                            radius_m
                        )
                    )

                    # --------------------------------------------
                    # RECONSTRUCTION FOR THIS INCIDENT
                    # --------------------------------------------

                    group_result = (
                        cw.reconstruct_state(
                            observations=
                                group_observations,

                            gis_context=
                                group_gis_context,

                            analysis_time=
                                analysis_time
                        )
                    )

                    # --------------------------------------------
                    # STORE INCIDENT RESULT
                    # --------------------------------------------

                    incident_results[
                        incident_id
                    ] = {
                        "incident_id":
                            incident_id,

                        "center_lat":
                            group_lat,

                        "center_lon":
                            group_lon,

                        "observation_count":
                            group[
                                "observation_count"
                            ],

                        "max_observation_distance_m":
                            group[
                                "max_observation_distance_m"
                            ],

                        "observation_ids": [
                            obs.get(
                                "observation_id"
                            )
                            for obs in group_observations
                        ],

                        "observations":
                            group_observations,

                        "gis_context":
                            group_gis_context,

                        "result":
                            group_result
                    }

                # ------------------------------------------------
                # STORE MULTI-INCIDENT RESULT
                # ------------------------------------------------

                results = {
                    "analysis_time":
                        analysis_time,

                    "incident_count":
                        len(incident_results),

                    "incidents":
                        incident_results
                }

                st.session_state[
                    "last_analysis"
                ] = results

                st.session_state[
                    "last_gis_context"
                ] = {
                    incident_id:
                        data["gis_context"]
                    for incident_id, data
                    in incident_results.items()
                }

                st.success(
                    f"Live reconstruction completed for "
                    f"{len(incident_results)} incident(s)."
                )

            except Exception as exc:

                st.error(
                    "Live reconstruction failed."
                )

                st.exception(exc)


'''

text = (
    text[:start]
    + new_block
    + text[end:]
)

APP_PATH.write_text(
    text,
    encoding="utf-8"
)

print("=" * 70)
print("CRISISWEAVE — MULTI-INCIDENT PATCH")
print("=" * 70)
print("✓ Live reconstruction block replaced")
print("✓ Existing reconstruct_state() untouched")
print("✓ Existing GIS logic untouched")
print("✓ Existing grouping function used")
print("✓ Per-incident GIS enabled")
print("✓ Per-incident reconstruction enabled")
print()
print("IMPORTANT:")
print("The existing result-display section still expects")
print("the old single-result structure.")
print("We will patch that next.")
```

```python
from pathlib import Path
from datetime import datetime
import py_compile

APP = Path(
    "/content/gdrive/MyDrive/"
    "CrisisWeave_Project/05_Streamlit/"
    "CrisisWeave_Live/live_app.py"
)

if not APP.exists():
    raise FileNotFoundError(
        f"live_app.py not found:\n{APP}"
    )

source = APP.read_text(
    encoding="utf-8"
)

print("=" * 78)
print("CRISISWEAVE — V6 INCIDENT MAP PATCH")
print("=" * 78)
print(
    f"File: {APP}"
)
print(
    f"Current lines: {len(source.splitlines())}"
)
print(
    f"Current characters: {len(source)}"
)


# ================================================================
# 1. BACKUP
# ================================================================

stamp = datetime.now().strftime(
    "%Y%m%d_%H%M%S"
)

backup = APP.with_name(
    f"live_app.py.backup_MAP_V6_{stamp}"
)

backup.write_text(
    source,
    encoding="utf-8"
)

print("\nBACKUP CREATED")
print(backup)


# ================================================================
# 2. REQUIRED IMPORTS
# ================================================================

if "import folium" not in source:

    source = (
        "import folium\n"
        + source
    )

    print(
        "✓ inserted: import folium"
    )

else:

    print(
        "✓ folium import already present"
    )


if (
    "from streamlit_folium import st_folium"
    not in source
):

    source = (
        "from streamlit_folium import st_folium\n"
        + source
    )

    print(
        "✓ inserted: "
        "from streamlit_folium import st_folium"
    )

else:

    print(
        "✓ streamlit_folium import already present"
    )


# ================================================================
# 3. PREVENT DUPLICATE MAP
# ================================================================

if (
    "### 🗺️ Incident Verification Map"
    in source
):

    raise RuntimeError(
        "Incident Verification Map already exists "
        "in live_app.py.\n"
        "No changes were made."
    )


# ================================================================
# 4. MAP UI BLOCK
#
# IMPORTANT:
# This block does NOT depend on the old
# 'RECONSTRUCTION RESULT' marker.
#
# It uses the current variables that already exist
# inside the per-incident display loop:
#
# incident_id
# incident_data
# incident_result
# center_lat
# center_lon
# ================================================================

map_block = r"""
        # ====================================================
        # V6 — INCIDENT VERIFICATION MAP
        # ====================================================

        st.markdown(
            "### 🗺️ Incident Verification Map"
        )

        # ----------------------------------------------------
        # Read existing incident data
        # ----------------------------------------------------

        map_gis_context = incident_data.get(
            "gis_context",
            {}
        )

        map_priority = incident_result.get(
            "priority",
            {}
        )

        map_selected = map_priority.get(
            "selected_next_priority"
        )

        # ----------------------------------------------------
        # Extract GIS features recursively.
        #
        # This deliberately does not assume one exact
        # gis_context structure.
        # ----------------------------------------------------

        map_features = []

        map_seen = set()


        def _cw_walk_map_features(value):

            if isinstance(value, dict):

                lat = None
                lon = None

                # --------------------------------------------
                # Latitude
                # --------------------------------------------

                for key in (
                    "latitude",
                    "lat",
                    "center_lat"
                ):

                    if value.get(key) is not None:

                        try:

                            lat = float(
                                value.get(key)
                            )

                            break

                        except Exception:

                            pass


                # --------------------------------------------
                # Longitude
                # --------------------------------------------

                for key in (
                    "longitude",
                    "lon",
                    "lng",
                    "center_lon"
                ):

                    if value.get(key) is not None:

                        try:

                            lon = float(
                                value.get(key)
                            )

                            break

                        except Exception:

                            pass


                # --------------------------------------------
                # Geographic feature
                # --------------------------------------------

                if (
                    lat is not None
                    and lon is not None
                    and -90 <= lat <= 90
                    and -180 <= lon <= 180
                ):

                    name = (
                        value.get("name")
                        or value.get("display_name")
                        or value.get("title")
                        or value.get("label")
                        or "Unnamed"
                    )

                    feature_type = (
                        value.get("feature_type")
                        or value.get("type")
                        or value.get("category")
                        or value.get("class")
                        or "GIS feature"
                    )

                    distance = (
                        value.get("distance_m")
                        if value.get("distance_m") is not None
                        else value.get("distance")
                    )

                    item_key = (
                        round(lat, 7),
                        round(lon, 7),
                        str(name),
                        str(feature_type)
                    )

                    if item_key not in map_seen:

                        map_seen.add(
                            item_key
                        )

                        map_features.append(
                            {
                                "name": str(name),
                                "feature_type": str(
                                    feature_type
                                ),
                                "latitude": lat,
                                "longitude": lon,
                                "distance_m": distance
                            }
                        )


                # --------------------------------------------
                # Continue recursively
                # --------------------------------------------

                for child in value.values():

                    _cw_walk_map_features(
                        child
                    )


            elif isinstance(
                value,
                (list, tuple)
            ):

                for child in value:

                    _cw_walk_map_features(
                        child
                    )


        _cw_walk_map_features(
            map_gis_context
        )


        # ====================================================
        # SELECTED NEXT VERIFICATION LOCATION
        # ====================================================

        selected_lat = None
        selected_lon = None


        if isinstance(
            map_selected,
            dict
        ):

            for key in (
                "latitude",
                "lat",
                "center_lat"
            ):

                if (
                    map_selected.get(key)
                    is not None
                ):

                    try:

                        selected_lat = float(
                            map_selected.get(key)
                        )

                        break

                    except Exception:

                        pass


            for key in (
                "longitude",
                "lon",
                "lng",
                "center_lon"
            ):

                if (
                    map_selected.get(key)
                    is not None
                ):

                    try:

                        selected_lon = float(
                            map_selected.get(key)
                        )

                        break

                    except Exception:

                        pass


        # ----------------------------------------------------
        # If selected location does not contain coordinates,
        # match it against the GIS candidate list by name.
        # ----------------------------------------------------

        if (
            map_selected
            and (
                selected_lat is None
                or selected_lon is None
            )
        ):

            selected_name_match = str(
                map_selected.get(
                    "name",
                    ""
                )
            ).strip().lower()


            for feature in map_features:

                feature_name = str(
                    feature.get(
                        "name",
                        ""
                    )
                ).strip().lower()


                if (
                    selected_name_match
                    and
                    selected_name_match
                    == feature_name
                ):

                    selected_lat = float(
                        feature["latitude"]
                    )

                    selected_lon = float(
                        feature["longitude"]
                    )

                    break


        # ====================================================
        # BUILD MAP
        # ====================================================

        try:

            map_center_lat = float(
                center_lat
            )

            map_center_lon = float(
                center_lon
            )


            incident_map = folium.Map(
                location=[
                    map_center_lat,
                    map_center_lon
                ],
                zoom_start=16,
                control_scale=True
            )


            # =================================================
            # INCIDENT CENTER
            # =================================================

            folium.Marker(
                location=[
                    map_center_lat,
                    map_center_lon
                ],

                tooltip=(
                    f"{incident_id} — Incident Center"
                ),

                popup=(
                    f"<b>{incident_id}</b><br>"
                    "Incident Center<br>"
                    f"Latitude: "
                    f"{map_center_lat:.6f}<br>"
                    f"Longitude: "
                    f"{map_center_lon:.6f}"
                ),

                icon=folium.Icon(
                    icon="warning-sign",
                    prefix="glyphicon",
                    color="red"
                )

            ).add_to(
                incident_map
            )


            # =================================================
            # INCIDENT WORKING RADIUS
            # =================================================

            folium.Circle(
                location=[
                    map_center_lat,
                    map_center_lon
                ],

                radius=100,

                tooltip=(
                    "Incident grouping radius: 100 m"
                ),

                fill=False

            ).add_to(
                incident_map
            )


            # =================================================
            # GIS CANDIDATES
            # =================================================

            for feature in map_features[:50]:

                try:

                    feature_lat = float(
                        feature["latitude"]
                    )

                    feature_lon = float(
                        feature["longitude"]
                    )

                except Exception:

                    continue


                feature_name = feature.get(
                    "name",
                    "Unnamed"
                )

                feature_type = feature.get(
                    "feature_type",
                    "GIS feature"
                )

                feature_distance = feature.get(
                    "distance_m"
                )


                # --------------------------------------------
                # Check whether this is selected
                # --------------------------------------------

                is_selected = False


                if (
                    selected_lat is not None
                    and selected_lon is not None
                ):

                    if (
                        abs(
                            feature_lat
                            - selected_lat
                        ) < 1e-6
                        and
                        abs(
                            feature_lon
                            - selected_lon
                        ) < 1e-6
                    ):

                        is_selected = True


                # --------------------------------------------
                # Popup
                # --------------------------------------------

                popup = (
                    f"<b>{feature_name}</b><br>"
                    f"Type: {feature_type}<br>"
                    f"Latitude: {feature_lat:.6f}<br>"
                    f"Longitude: {feature_lon:.6f}"
                )


                if feature_distance is not None:

                    try:

                        popup += (
                            "<br>Distance: "
                            f"{float(feature_distance):.1f} m"
                        )

                    except Exception:

                        pass


                # --------------------------------------------
                # Candidate marker
                # --------------------------------------------

                folium.Marker(
                    location=[
                        feature_lat,
                        feature_lon
                    ],

                    tooltip=(
                        "★ "
                        if is_selected
                        else ""
                    )
                    + str(feature_name),

                    popup=popup,

                    icon=folium.Icon(
                        icon=(
                            "star"
                            if is_selected
                            else "info-sign"
                        ),

                        prefix="glyphicon",

                        color=(
                            "green"
                            if is_selected
                            else "blue"
                        )
                    )

                ).add_to(
                    incident_map
                )


            # =================================================
            # NEXT VERIFICATION LOCATION
            # =================================================

            if (
                map_selected
                and
                selected_lat is not None
                and
                selected_lon is not None
            ):

                selected_name = map_selected.get(
                    "name",
                    "Next verification location"
                )

                selected_type = map_selected.get(
                    "feature_type",
                    map_selected.get(
                        "type",
                        "GIS feature"
                    )
                )


                # --------------------------------------------
                # Red flag marker
                # --------------------------------------------

                folium.Marker(

                    location=[
                        selected_lat,
                        selected_lon
                    ],

                    tooltip=(
                        "NEXT VERIFICATION LOCATION — "
                        f"{selected_name}"
                    ),

                    popup=(
                        "<b>Next Verification Location</b><br>"
                        f"{selected_name}<br>"
                        f"Type: {selected_type}<br>"
                        f"Latitude: {selected_lat:.6f}<br>"
                        f"Longitude: {selected_lon:.6f}"
                    ),

                    icon=folium.Icon(
                        icon="flag",
                        prefix="glyphicon",
                        color="red"
                    )

                ).add_to(
                    incident_map
                )


                # --------------------------------------------
                # Incident → verification route
                # --------------------------------------------

                folium.PolyLine(

                    locations=[
                        [
                            map_center_lat,
                            map_center_lon
                        ],

                        [
                            selected_lat,
                            selected_lon
                        ]
                    ],

                    tooltip=(
                        "Incident → next verification location"
                    ),

                    weight=4

                ).add_to(
                    incident_map
                )


            # =================================================
            # FIT MAP TO ALL RELEVANT LOCATIONS
            # =================================================

            map_bounds = [
                [
                    map_center_lat,
                    map_center_lon
                ]
            ]


            for feature in map_features[:50]:

                try:

                    map_bounds.append(
                        [
                            float(
                                feature["latitude"]
                            ),
                            float(
                                feature["longitude"]
                            )
                        ]
                    )

                except Exception:

                    pass


            if (
                selected_lat is not None
                and selected_lon is not None
            ):

                map_bounds.append(
                    [
                        selected_lat,
                        selected_lon
                    ]
                )


            if len(map_bounds) > 1:

                incident_map.fit_bounds(
                    map_bounds,
                    padding=(
                        20,
                        20
                    )
                )


            # =================================================
            # RENDER STREAMLIT MAP
            # =================================================

            st_folium(

                incident_map,

                width=None,

                height=520,

                returned_objects=[],

                key=(
                    f"cw_incident_verification_map_"
                    f"{incident_id}"
                )
            )


            st.caption(
                f"Incident center + "
                f"{min(len(map_features), 50)} "
                "nearby GIS feature(s). "
                "Red flag = next verification location."
            )


        except Exception as map_error:

            st.error(
                "Incident map could not be rendered."
            )

            st.exception(
                map_error
            )

"""

# ================================================================
# 5. FIND OPERATIONAL RECOMMENDATION
#
# We DO NOT search for:
#     RECONSTRUCTION RESULT
#
# Instead, we use the actual heading that exists in the current
# file and insert the map immediately before it.
# ================================================================

anchor_candidates = [

    '        st.markdown(\n'
    '            "### Operational Recommendation"\n'
    '        )',

    '            "### Operational Recommendation"',

    '"### Operational Recommendation"'

]


insert_at = -1
used_anchor = None


for anchor in anchor_candidates:

    position = source.rfind(
        anchor
    )

    if position != -1:

        insert_at = position
        used_anchor = anchor

        break


if insert_at == -1:

    raise RuntimeError(
        "Could not find the existing "
        "'### Operational Recommendation' "
        "heading.\n\n"
        "No changes were made."
    )


# ================================================================
# 6. SAFETY CHECK
# ================================================================

state_markers = [

    "# LIVE RESULT SUMMARY",

    "# CURRENT LIVE STATE",

    "### Current Live State",

    "Current Live State"

]


state_positions = []


for marker in state_markers:

    position = source.rfind(
        marker
    )

    if position != -1:

        state_positions.append(
            position
        )


if state_positions:

    latest_state_position = max(
        state_positions
    )

    if insert_at < latest_state_position:

        raise RuntimeError(
            "The located Operational Recommendation "
            "appears before Current Live State.\n"
            "No changes were made."
        )


# ================================================================
# 7. INSERT MAP
# ================================================================

source = (

    source[:insert_at]

    + map_block

    + "\n"

    + source[insert_at:]

)


APP.write_text(
    source,
    encoding="utf-8"
)


# ================================================================
# 8. PYTHON COMPILE CHECK
# ================================================================

py_compile.compile(
    str(APP),
    doraise=True
)


# ================================================================
# 9. RESULT
# ================================================================

print("\n" + "=" * 78)
print("PATCH RESULT")
print("=" * 78)

print(
    "✓ Incident Verification Map inserted"
)

print(
    "✓ Existing multi-incident reconstruction untouched"
)

print(
    "✓ Existing Next Verification Location untouched"
)

print(
    "✓ Existing Operational Recommendation untouched"
)

print(
    "✓ folium import:",
    "import folium" in source
)

print(
    "✓ st_folium import:",
    "from streamlit_folium import st_folium"
    in source
)

print(
    "✓ Map UI present:",
    "### 🗺️ Incident Verification Map"
    in source
)

print(
    "✓ New lines:",
    len(source.splitlines())
)

print(
    "✓ New characters:",
    len(source)
)

print(
    "✓ Python compilation passed"
)

print("\n" + "=" * 78)
print("PATCH COMPLETE")
print("=" * 78)

print(
    "\nBackup available at:"
)
print(backup)
```

```python
# ============================================================================
# CRISISWEAVE — FINAL SELECTED INCIDENT UI + WORKING VERIFICATION MAP
# ============================================================================
# PURPOSE:
#   1. Keep multi-incident reconstruction intact.
#   2. Show Incident 1 / Incident 2 / Incident 3 selector.
#   3. Display ONLY the selected incident's analysis.
#   4. Put the verification map directly under Next Verification Location.
#   5. Use the original working Folium-style map:
#        🔴 Incident center
#        ⭐ Next verification location
#        🟢 School
#        ➕ Red Hospital
#        🟠 Bridge
#        💧 Blue water resource
#        ⚪ Gray other useful GIS feature
#   6. Display recommendation as normal UI, NOT raw JSON.
#   7. Do NOT modify live_engine.py.
#   8. Do NOT depend on local incident_id outside the incident loop.
# ============================================================================

from pathlib import Path
from datetime import datetime

APP = Path(
    "/content/gdrive/MyDrive/"
    "CrisisWeave_Project/05_Streamlit/"
    "CrisisWeave_Live/live_app.py"
)

print("=" * 78)
print("CRISISWEAVE — FINAL SELECTED INCIDENT UI + WORKING MAP")
print("=" * 78)

if not APP.exists():
    raise FileNotFoundError(
        f"live_app.py not found:\n{APP}"
    )

source = APP.read_text(encoding="utf-8")

print(f"✓ live_app.py found")
print(f"  Current lines      : {len(source.splitlines())}")
print(f"  Current characters : {len(source)}")


# ============================================================================
# BACKUP
# ============================================================================

stamp = datetime.now().strftime("%Y%m%d_%H%M%S")

backup = APP.with_name(
    APP.name +
    f".backup_FINAL_SELECTED_INCIDENT_UI_{stamp}"
)

backup.write_text(
    source,
    encoding="utf-8"
)

print()
print("✓ Backup created:")
print(f"  {backup}")


# ============================================================================
# FIND CURRENT RESULT DISPLAY
# ============================================================================
#
# We intentionally replace from the CURRENT LIVE STATE section to the FOOTER.
#
# This avoids searching for fragile strings such as:
#   color="green"
#   icon="education"
#   Incident Verification Map
#   RECOMMENDATION boundary
#
# Those were the reason previous patches repeatedly failed.
# ============================================================================

start_candidates = [
    "# LIVE RESULT SUMMARY",
    "# CURRENT LIVE STATE",
    "# LIVE RESULT DISPLAY",
    "# RESULT DISPLAY",
]

start_pos = -1
start_token = None

for token in start_candidates:
    pos = source.find(token)
    if pos != -1:
        start_pos = pos
        start_token = token
        break


# If the exact comment changed, locate the first Current Live State heading.
if start_pos == -1:

    marker = 'st.subheader(\n        "6. Current Live State"'

    pos = source.find(marker)

    if pos != -1:
        # Walk backwards to the nearest section comment.
        section_pos = source.rfind(
            "# ============================================================",
            0,
            pos
        )

        if section_pos != -1:
            start_pos = section_pos
            start_token = "CURRENT LIVE STATE section"

if start_pos == -1:

    raise RuntimeError(
        "Could not safely locate the current result-display section.\n"
        "NO CHANGES WERE WRITTEN.\n\n"
        "Run this diagnostic first:\n"
        "print('\\n'.join(f'{i+1}: {x}' for i,x in "
        "enumerate(source.splitlines()) if "
        "'Current Live State' in x or "
        "'LIVE RESULT' in x))"
    )


# ============================================================================
# FOOTER
# ============================================================================

footer_tokens = [
    "# ============================================================\n"
    "# FOOTER",

    "# ============================================================\r\n"
    "# FOOTER",
]

footer_pos = -1

for token in footer_tokens:

    pos = source.find(
        token,
        start_pos + 1
    )

    if pos != -1:
        footer_pos = pos
        break


if footer_pos == -1:

    # More tolerant fallback.
    footer_heading = source.find(
        "# FOOTER",
        start_pos + 1
    )

    if footer_heading != -1:
        footer_pos = source.rfind(
            "# ============================================================",
            start_pos + 1,
            footer_heading + 1
        )


if footer_pos == -1:

    raise RuntimeError(
        "Could not safely locate the footer boundary.\n"
        "NO CHANGES WERE WRITTEN."
    )


print()
print("✓ Current result-display section located")
print(f"  Start character : {start_pos}")
print(f"  Footer character: {footer_pos}")


# ============================================================================
# NEW RESULT DISPLAY
# ============================================================================

new_result_section = r'''
# ============================================================
# LIVE RESULT SUMMARY
# ============================================================

results = st.session_state.get(
    "last_analysis"
)

if results is not None:

    st.divider()

    st.subheader(
        "6. Current Live State"
    )

    incident_count = int(
        results.get(
            "incident_count",
            0
        )
    )

    incidents = results.get(
        "incidents",
        {}
    )

    if not isinstance(
        incidents,
        dict
    ):
        incidents = {}

    st.info(
        f"Live reconstruction contains "
        f"{incident_count} incident(s)."
    )

    # ========================================================
    # INCIDENT SELECTOR
    # ========================================================
    #
    # Only one incident is displayed at a time.
    # This prevents Incident 1, Incident 2, Incident 3
    # from becoming one huge page.
    # ========================================================

    incident_ids = list(
        incidents.keys()
    )

    if incident_ids:

        if (
            "selected_incident_id"
            not in st.session_state
            or
            st.session_state[
                "selected_incident_id"
            ] not in incident_ids
        ):

            st.session_state[
                "selected_incident_id"
            ] = incident_ids[0]

        st.markdown(
            "### Select Incident"
        )

        selected_incident_id = st.radio(
            "Incident",
            incident_ids,
            index=incident_ids.index(
                st.session_state[
                    "selected_incident_id"
                ]
            ),
            horizontal=True,
            key="incident_selector"
        )

        st.session_state[
            "selected_incident_id"
        ] = selected_incident_id

        # ----------------------------------------------------
        # SELECTED INCIDENT
        # ----------------------------------------------------

        incident_id = (
            selected_incident_id
        )

        incident_data = incidents.get(
            incident_id,
            {}
        )

        st.divider()

        st.markdown(
            f"## 🚨 {incident_id}"
        )

        # ====================================================
        # INCIDENT METADATA
        # ====================================================

        center_lat = incident_data.get(
            "center_lat"
        )

        center_lon = incident_data.get(
            "center_lon"
        )

        observation_count = incident_data.get(
            "observation_count",
            0
        )

        max_distance = incident_data.get(
            "max_observation_distance_m",
            0.0
        )

        metric1, metric2, metric3 = st.columns(
            3
        )

        with metric1:

            st.metric(
                "Observations",
                observation_count
            )

        with metric2:

            st.metric(
                "Center Latitude",
                (
                    f"{float(center_lat):.6f}"
                    if center_lat is not None
                    else "UNKNOWN"
                )
            )

        with metric3:

            st.metric(
                "Cluster Radius",
                (
                    f"{float(max_distance):.1f} m"
                    if max_distance is not None
                    else "UNKNOWN"
                )
            )

        if (
            center_lat is not None
            and center_lon is not None
        ):

            st.caption(
                f"Incident center: "
                f"{float(center_lat):.6f}, "
                f"{float(center_lon):.6f}"
            )

        # ====================================================
        # OBSERVATION IDS
        # ====================================================

        observation_ids = incident_data.get(
            "observation_ids",
            []
        )

        if observation_ids:

            st.write(
                "Observation IDs:",
                ", ".join(
                    str(x)
                    for x in observation_ids
                    if x is not None
                )
            )

        # ====================================================
        # RECONSTRUCTION RESULT
        # ====================================================

        incident_result = incident_data.get(
            "result",
            {}
        )

        if not isinstance(
            incident_result,
            dict
        ):
            incident_result = {}

        fusion = incident_result.get(
            "fusion",
            {}
        )

        temporal = incident_result.get(
            "temporal",
            {}
        )

        conflict = incident_result.get(
            "conflict",
            {}
        )

        verification = incident_result.get(
            "verification",
            {}
        )

        dynamic = incident_result.get(
            "dynamic_state",
            {}
        )

        recommendation = incident_result.get(
            "recommendation",
            {}
        )

        priority = incident_result.get(
            "priority",
            {}
        )

        if not isinstance(
            priority,
            dict
        ):
            priority = {}

        # ====================================================
        # RECONSTRUCTION SUMMARY
        # ====================================================

        st.markdown(
            "### Reconstruction Summary"
        )

        metric1, metric2, metric3, metric4 = st.columns(
            4
        )

        with metric1:

            st.metric(
                "Fused Severity",
                fusion.get(
                    "fused_severity",
                    "UNKNOWN"
                )
            )

        with metric2:

            try:

                confidence_value = float(
                    fusion.get(
                        "confidence",
                        0
                    )
                )

                confidence_text = (
                    f"{confidence_value * 100:.1f}%"
                )

            except Exception:

                confidence_text = "UNKNOWN"

            st.metric(
                "Fusion Confidence",
                confidence_text
            )

        with metric3:

            st.metric(
                "Temporal Trend",
                temporal.get(
                    "trend",
                    "UNKNOWN"
                )
            )

        with metric4:

            st.metric(
                "Verification",
                verification.get(
                    "status",
                    "UNKNOWN"
                )
            )

        # ====================================================
        # TEMPORAL + CONFLICT
        # ====================================================

        detail1, detail2 = st.columns(
            2
        )

        with detail1:

            st.markdown(
                "### Temporal State"
            )

            st.write(
                "Observations:",
                temporal.get(
                    "observation_count",
                    observation_count
                )
            )

            st.write(
                "Initial severity:",
                temporal.get(
                    "initial_severity",
                    "UNKNOWN"
                )
            )

            st.write(
                "Latest severity:",
                temporal.get(
                    "latest_severity",
                    "UNKNOWN"
                )
            )

            st.write(
                "Trend:",
                temporal.get(
                    "trend",
                    "UNKNOWN"
                )
            )

            st.write(
                "Progression:",
                temporal.get(
                    "progression_direction",
                    "UNKNOWN"
                )
            )

        with detail2:

            st.markdown(
                "### Evidence Conflict"
            )

            st.write(
                "Status:",
                conflict.get(
                    "status",
                    "UNKNOWN"
                )
            )

            st.write(
                "Overall conflict:",
                conflict.get(
                    "overall_conflict",
                    "UNKNOWN"
                )
            )

            st.write(
                "Difference:",
                conflict.get(
                    "difference",
                    "UNKNOWN"
                )
            )

            st.write(
                "Humanitarian evidence:",
                conflict.get(
                    "humanitarian_evidence",
                    "UNKNOWN"
                )
            )

        # ====================================================
        # RELIABILITY
        # ====================================================

        reliability = incident_result.get(
            "reliability",
            {}
        )

        st.markdown(
            "### Reliability"
        )

        try:

            mean_reliability = float(
                reliability.get(
                    "mean_reliability",
                    0
                )
            )

        except Exception:

            mean_reliability = 0.0

        st.write(
            "Mean reliability:",
            round(
                mean_reliability,
                4
            )
        )

        # ====================================================
        # DYNAMIC DISASTER STATE
        # ====================================================

        st.markdown(
            "### Dynamic Disaster State"
        )

        current_state = dynamic.get(
            "current_state",
            {}
        )

        if not isinstance(
            current_state,
            dict
        ):
            current_state = {}

        st.write(
            "Severity:",
            current_state.get(
                "severity",
                "UNKNOWN"
            )
        )

        st.write(
            "Confidence:",
            current_state.get(
                "confidence",
                0
            )
        )

        st.write(
            "Uncertainty:",
            current_state.get(
                "uncertainty",
                "UNKNOWN"
            )
        )

        st.write(
            "Operational risk:",
            dynamic.get(
                "operational_risk",
                "UNKNOWN"
            )
        )

        # ====================================================
        # NEXT VERIFICATION LOCATION
        # ====================================================

        selected_next = priority.get(
            "selected_next_priority"
        )

        st.markdown(
            "### 🎯 Next Verification Location"
        )

        if selected_next:

            next_name = selected_next.get(
                "name",
                "Unnamed location"
            )

            next_type = selected_next.get(
                "feature_type",
                "Unknown"
            )

            next_distance = selected_next.get(
                "distance_m"
            )

            next_score = selected_next.get(
                "priority_score"
            )

            direction_alignment = selected_next.get(
                "direction_alignment"
            )

            st.success(
                f"{next_name} — {next_type}"
            )

            m1, m2, m3 = st.columns(
                3
            )

            with m1:

                if next_distance is not None:

                    st.metric(
                        "Distance",
                        f"{float(next_distance):.1f} m"
                    )

                else:

                    st.metric(
                        "Distance",
                        "UNKNOWN"
                    )

            with m2:

                if next_score is not None:

                    st.metric(
                        "Priority Score",
                        f"{float(next_score):.3f}"
                    )

                else:

                    st.metric(
                        "Priority Score",
                        "UNKNOWN"
                    )

            with m3:

                if direction_alignment is not None:

                    st.metric(
                        "Direction Alignment",
                        f"{float(direction_alignment):.2f}"
                    )

                else:

                    st.metric(
                        "Direction Alignment",
                        "UNKNOWN"
                    )

            st.markdown(
                "**Why this location?**"
            )

            st.write(
                "The location is selected from eligible "
                "nearby GIS candidates using incident "
                "proximity, directional alignment, "
                "water proximity, and feature criticality."
            )

            basis = selected_next.get(
                "selection_basis"
            )

            if isinstance(
                basis,
                dict
            ) and basis:

                st.write(
                    "Selection basis:"
                )

                basis_items = []

                for key, value in basis.items():

                    label = str(
                        key
                    ).replace(
                        "_",
                        " "
                    ).title()

                    if isinstance(
                        value,
                        float
                    ):

                        value_text = (
                            f"{value:.3f}"
                        )

                    else:

                        value_text = str(
                            value
                        )

                    basis_items.append(
                        f"**{label}:** {value_text}"
                    )

                st.markdown(
                    "  \n".join(
                        basis_items
                    )
                )

        else:

            st.info(
                "No next verification location selected."
            )

        # ====================================================
        # INCIDENT VERIFICATION MAP
        #
        # IMPORTANT:
        # The map is intentionally inside the selected incident
        # block and directly follows Next Verification Location.
        #
        # No separate 'Incident Verification Map' title.
        # ====================================================

        try:

            gis_context = incident_data.get(
                "gis_context",
                {}
            )

            if not isinstance(
                gis_context,
                dict
            ):
                gis_context = {}

            verification_features = gis_context.get(
                "features",
                []
            )

            if not isinstance(
                verification_features,
                list
            ):
                verification_features = []

            if (
                center_lat is not None
                and center_lon is not None
            ):

                verification_map = folium.Map(
                    location=[
                        float(center_lat),
                        float(center_lon)
                    ],
                    zoom_start=15,
                    control_scale=True
                )

                # ------------------------------------------------
                # INCIDENT CENTER
                # ------------------------------------------------

                folium.Marker(
                    [
                        float(center_lat),
                        float(center_lon)
                    ],
                    tooltip=(
                        f"🔴 {incident_id} — Incident Center"
                    ),
                    popup=(
                        f"<b>🔴 Incident Center</b><br>"
                        f"<b>Incident:</b> {incident_id}<br>"
                        f"<b>Latitude:</b> {float(center_lat):.6f}<br>"
                        f"<b>Longitude:</b> {float(center_lon):.6f}"
                    ),
                    icon=folium.Icon(
                        color="red",
                        icon="warning-sign"
                    )
                ).add_to(
                    verification_map
                )

                # ------------------------------------------------
                # GIS FEATURE ICONS
                # ------------------------------------------------
                #
                # These are the SAME semantic categories used
                # by the original working verification map.
                #
                # School   -> green education
                # Hospital -> red plus
                # Bridge   -> orange road
                # Water    -> blue tint
                # Other    -> gray info-sign
                # ------------------------------------------------

                for feature in verification_features:

                    if not isinstance(
                        feature,
                        dict
                    ):
                        continue

                    lat = feature.get(
                        "latitude"
                    )

                    lon = feature.get(
                        "longitude"
                    )

                    if (
                        lat is None
                        or lon is None
                    ):
                        continue

                    try:

                        lat = float(
                            lat
                        )

                        lon = float(
                            lon
                        )

                    except Exception:

                        continue

                    feature_type = str(
                        feature.get(
                            "feature_type",
                            "Other"
                        )
                    )

                    name = str(
                        feature.get(
                            "name",
                            "Unnamed"
                        )
                    )

                    distance = feature.get(
                        "distance_m"
                    )

                    # Normalize feature type.
                    feature_type_lower = (
                        feature_type
                        .strip()
                        .lower()
                    )

                    # ------------------------------------------------
                    # SCHOOL
                    # ------------------------------------------------

                    if (
                        feature_type_lower
                        in {
                            "school",
                            "schools",
                            "educational institution",
                            "education"
                        }
                    ):

                        marker_color = "green"
                        marker_icon = "education"
                        display_type = "School"

                    # ------------------------------------------------
                    # HOSPITAL
                    # ------------------------------------------------

                    elif (
                        feature_type_lower
                        in {
                            "hospital",
                            "hospitals",
                            "clinic",
                            "healthcare"
                        }
                    ):

                        marker_color = "red"
                        marker_icon = "plus"
                        display_type = "Hospital"

                    # ------------------------------------------------
                    # BRIDGE
                    # ------------------------------------------------

                    elif (
                        feature_type_lower
                        in {
                            "bridge",
                            "bridges"
                        }
                    ):

                        marker_color = "orange"
                        marker_icon = "road"
                        display_type = "Bridge"

                    # ------------------------------------------------
                    # WATER
                    # ------------------------------------------------

                    elif (
                        feature_type_lower
                        in {
                            "waterbody",
                            "water body",
                            "water",
                            "water resource",
                            "water resources",
                            "reservoir",
                            "reservoirs",
                            "lake",
                            "river",
                            "canal",
                            "pond"
                        }
                        or
                        "water" in feature_type_lower
                    ):

                        marker_color = "blue"
                        marker_icon = "tint"
                        display_type = "Water Resource"

                    # ------------------------------------------------
                    # OTHER USEFUL GIS FEATURES
                    # ------------------------------------------------

                    else:

                        marker_color = "gray"
                        marker_icon = "info-sign"
                        display_type = feature_type

                    distance_text = (
                        f"{float(distance):.1f} m"
                        if distance is not None
                        else "Unknown"
                    )

                    popup_text = (
                        f"<b>{name}</b><br>"
                        f"<b>Type:</b> {display_type}<br>"
                        f"<b>Distance:</b> {distance_text}"
                    )

                    folium.Marker(
                        [
                            lat,
                            lon
                        ],
                        tooltip=(
                            f"{display_type}: {name}"
                        ),
                        popup=popup_text,
                        icon=folium.Icon(
                            color=marker_color,
                            icon=marker_icon
                        )
                    ).add_to(
                        verification_map
                    )

                # ------------------------------------------------
                # NEXT VERIFICATION LOCATION
                # ------------------------------------------------
                #
                # This is drawn LAST so the purple star is visible
                # above other GIS markers when locations overlap.
                # ------------------------------------------------

                if selected_next:

                    next_lat = selected_next.get(
                        "latitude"
                    )

                    next_lon = selected_next.get(
                        "longitude"
                    )

                    if (
                        next_lat is not None
                        and next_lon is not None
                    ):

                        try:

                            next_lat = float(
                                next_lat
                            )

                            next_lon = float(
                                next_lon
                            )

                            next_name = selected_next.get(
                                "name",
                                "Unnamed location"
                            )

                            next_type = selected_next.get(
                                "feature_type",
                                "Unknown"
                            )

                            next_distance = selected_next.get(
                                "distance_m",
                                0
                            )

                            next_score = selected_next.get(
                                "priority_score",
                                0
                            )

                            folium.Marker(
                                [
                                    next_lat,
                                    next_lon
                                ],
                                tooltip=(
                                    "⭐ NEXT VERIFICATION — "
                                    f"{next_name}"
                                ),
                                popup=(
                                    "<b>⭐ NEXT VERIFICATION</b><br>"
                                    f"<b>Location:</b> {next_name}<br>"
                                    f"<b>Type:</b> {next_type}<br>"
                                    f"<b>Distance:</b> "
                                    f"{next_distance} m<br>"
                                    f"<b>Priority Score:</b> "
                                    f"{next_score}"
                                ),
                                icon=folium.Icon(
                                    color="purple",
                                    icon="star"
                                )
                            ).add_to(
                                verification_map
                            )

                        except Exception:
                            pass

                # ------------------------------------------------
                # MAP LEGEND
                # ------------------------------------------------

                legend_html = """
                <div style="
                    position: fixed;
                    bottom: 25px;
                    left: 25px;
                    z-index: 9999;
                    background-color: white;
                    border: 2px solid #777;
                    border-radius: 6px;
                    padding: 10px 12px;
                    font-size: 13px;
                    box-shadow: 0 2px 6px rgba(0,0,0,0.25);
                ">
                    <b>CrisisWeave Verification</b><br>
                    🔴 Incident Center<br>
                    ⭐ Next Verification<br>
                    🟢 School<br>
                    ➕ Hospital<br>
                    🟠 Bridge<br>
                    💧 Water Resource<br>
                    ⚪ Other GIS Feature
                </div>
                """

                verification_map.get_root().html.add_child(
                    folium.Element(
                        legend_html
                    )
                )

                st_folium(
                    verification_map,
                    width=None,
                    height=600,
                    key=(
                        f"verification_map_"
                        f"{incident_id}"
                    ),
                    returned_objects=[],
                    use_container_width=True
                )

            else:

                st.warning(
                    "Incident coordinates are unavailable, "
                    "so the verification map cannot be rendered."
                )

        except Exception as map_exc:

            st.error(
                "Verification map could not be rendered."
            )

            st.exception(
                map_exc
            )

        # ====================================================
        # SPATIAL VERIFICATION
        # ====================================================

        st.markdown(
            "### Spatial Verification"
        )

        verification_status = verification.get(
            "status",
            "UNKNOWN"
        )

        verification_score = verification.get(
            "score"
        )

        if verification_score is None:

            verification_score = verification.get(
                "verification_score"
            )

        if verification_score is not None:

            try:

                verification_percent = (
                    float(
                        verification_score
                    ) * 100
                )

                verification_percent_text = (
                    f"{verification_percent:.1f}%"
                )

            except Exception:

                verification_percent_text = (
                    str(
                        verification_score
                    )
                )

        else:

            verification_percent_text = (
                "UNKNOWN"
            )

        vm1, vm2 = st.columns(
            2
        )

        with vm1:

            st.metric(
                "Verification Status",
                verification_status
            )

        with vm2:

            st.metric(
                "Verification Score",
                verification_percent_text
            )

        if center_lat is not None:

            st.write(
                "Incident latitude:",
                float(center_lat)
            )

        if center_lon is not None:

            st.write(
                "Incident longitude:",
                float(center_lon)
            )

        # ====================================================
        # OPERATIONAL RECOMMENDATION
        # ====================================================

        st.markdown(
            "### Operational Recommendation"
        )

        decision = recommendation.get(
            "decision",
            "INSUFFICIENT EVIDENCE"
        )

        explanation = recommendation.get(
            "explanation",
            ""
        )

        st.warning(
            decision
        )

        if explanation:

            st.write(
                explanation
            )

        # ----------------------------------------------------
        # RECOMMENDATION EVIDENCE
        # ----------------------------------------------------

        evidence = recommendation.get(
            "evidence",
            {}
        )

        if isinstance(
            evidence,
            dict
        ) and evidence:

            st.markdown(
                "#### Recommendation Evidence"
            )

            ev1, ev2, ev3 = st.columns(
                3
            )

            with ev1:

                st.write(
                    "**Fused severity:**",
                    evidence.get(
                        "fused_severity",
                        "N/A"
                    )
                )

                st.write(
                    "**Latest severity:**",
                    evidence.get(
                        "latest_severity",
                        "N/A"
                    )
                )

                st.write(
                    "**Temporal trend:**",
                    evidence.get(
                        "temporal_trend",
                        "N/A"
                    )
                )

            with ev2:

                fusion_conf = evidence.get(
                    "fusion_confidence"
                )

                if fusion_conf is not None:

                    try:

                        fusion_conf_text = (
                            f"{float(fusion_conf) * 100:.1f}%"
                        )

                    except Exception:

                        fusion_conf_text = str(
                            fusion_conf
                        )

                else:

                    fusion_conf_text = "N/A"

                st.write(
                    "**Fusion confidence:**",
                    fusion_conf_text
                )

                st.write(
                    "**Uncertainty:**",
                    evidence.get(
                        "uncertainty",
                        "N/A"
                    )
                )

                st.write(
                    "**Verification:**",
                    evidence.get(
                        "verification_status",
                        verification_status
                    )
                )

            with ev3:

                rec_score = evidence.get(
                    "verification_score"
                )

                if rec_score is not None:

                    try:

                        rec_score_text = (
                            f"{float(rec_score) * 100:.1f}%"
                        )

                    except Exception:

                        rec_score_text = str(
                            rec_score
                        )

                else:

                    rec_score_text = "N/A"

                st.write(
                    "**Verification score:**",
                    rec_score_text
                )

                st.write(
                    "**Conflict status:**",
                    evidence.get(
                        "conflict_status",
                        "N/A"
                    )
                )

        # ====================================================
        # INCIDENT OBSERVATIONS
        # ====================================================

        incident_observations = incident_data.get(
            "observations",
            []
        )

        if incident_observations:

            with st.expander(
                f"View {incident_id} observations"
            ):

                observation_rows = []

                for obs in incident_observations:

                    if not isinstance(
                        obs,
                        dict
                    ):
                        continue

                    prediction = obs.get(
                        "prediction",
                        {}
                    )

                    if not isinstance(
                        prediction,
                        dict
                    ):
                        prediction = {}

                    observation_rows.append(
                        {
                            "Observation ID":
                                obs.get(
                                    "observation_id",
                                    "N/A"
                                ),

                            "Timestamp":
                                obs.get(
                                    "timestamp",
                                    "N/A"
                                ),

                            "Latitude":
                                obs.get(
                                    "latitude",
                                    "N/A"
                                ),

                            "Longitude":
                                obs.get(
                                    "longitude",
                                    "N/A"
                                ),

                            "Severity":
                                prediction.get(
                                    "severity",
                                    "N/A"
                                ),

                            "Category":
                                prediction.get(
                                    "category",
                                    "N/A"
                                )
                        }
                    )

                if observation_rows:

                    st.dataframe(
                        pd.DataFrame(
                            observation_rows
                        ),
                        use_container_width=True,
                        hide_index=True
                    )

    else:

        st.warning(
            "No incident results are available."
        )


# ============================================================
# FOOTER
# ============================================================

st.caption(
    "CrisisWeave Live — runtime evidence reconstruction. "
    "Saved result files are not used as analysis inputs."
)
'''


# ============================================================================
# WRITE
# ============================================================================

patched_source = (
    source[:start_pos]
    +
    new_result_section
    +
    "\n\n"
    +
    source[footer_pos:]
)

APP.write_text(
    patched_source,
    encoding="utf-8"
)

print()
print("=" * 78)
print("PATCH WRITTEN")
print("=" * 78)

print(
    f"✓ New lines       : "
    f"{len(patched_source.splitlines())}"
)

print(
    f"✓ New characters  : "
    f"{len(patched_source)}"
)


# ============================================================================
# COMPILE — BEFORE RUNNING STREAMLIT
# ============================================================================

import py_compile

try:

    py_compile.compile(
        str(APP),
        doraise=True
    )

    print(
        "✓ Python compilation passed"
    )

except Exception as exc:

    # Restore backup immediately.
    APP.write_text(
        source,
        encoding="utf-8"
    )

    print()
    print(
        "❌ Compilation failed."
    )

    print(
        "✓ Original file restored."
    )

    raise


# ============================================================================
# FINAL STRUCTURAL CHECKS
# ============================================================================

final_source = APP.read_text(
    encoding="utf-8"
)

required_checks = {
    "selected incident selector":
        'key="incident_selector"',

    "incident center marker":
        'icon="warning-sign"',

    "next verification marker":
        'icon="star"',

    "school marker":
        'marker_icon = "education"',

    "hospital marker":
        'marker_icon = "plus"',

    "bridge marker":
        'marker_icon = "road"',

    "water marker":
        'marker_icon = "tint"',

    "gray marker":
        'marker_color = "gray"',

    "map renderer":
        "st_folium(",

    "recommendation":
        "### Operational Recommendation",

    "recommendation evidence":
        "Recommendation Evidence",
}

print()
print(
    "FINAL CHECKS"
)

for label, token in required_checks.items():

    if token in final_source:

        print(
            f"✓ {label}"
        )

    else:

        # Restore if the structural check fails.
        APP.write_text(
            source,
            encoding="utf-8"
        )

        raise RuntimeError(
            f"Final safety check failed: "
            f"{label}\n"
            f"Token not found: {token}\n\n"
            f"Original file restored."
        )


print()
print("=" * 78)
print("PATCH COMPLETE — SAFE TO RUN")
print("=" * 78)

print()
print("Backup:")
print(backup)

print()
print("Run Streamlit ONLY after this cell finishes successfully.")
print()
print("Expected UI:")
print("  Incident 1 | Incident 2 | Incident 3")
print("       ↓")
print("  Selected incident analysis")
print("       ↓")
print("  🎯 Next Verification Location")
print("       ↓")
print("  🗺️ map")
print("       ↓")
print("  🔴 incident")
print("  ⭐ next verification")
print("  🟢 school")
print("  ➕ hospital")
print("  🟠 bridge")
print("  💧 water resource")
print("  ⚪ other GIS feature")
print("       ↓")
print("  Spatial Verification")
print("       ↓")
print("  Operational Recommendation")
print("       ↓")
print("  Recommendation Evidence")
print()
```

```python
# =============================================================================
# CRISISWEAVE — SAFE FINAL MULTI-INCIDENT MAP + NEXT VERIFICATION PATCH
# =============================================================================
#
# RUN THIS ONE CELL IN COLAB.
#
# FIXES:
#   1. No undefined incident_id problem
#   2. No "Incident Verification Map" heading in UI
#   3. Every incident gets its own unique map key
#   4. 🔴 red warning = incident center
#   5. ⭐ purple star = next verification location
#   6. 🟢 green = schools
#   7. ➕ red = hospitals
#   8. 🟠 bridge icon = bridges
#   9. 💧 blue = water resources
#  10. ⚪ gray = other GIS features
#  11. Roads are NOT displayed
#  12. Selected next location is shown ONLY as the purple star
#  13. Next location is proximity-first:
#          nearest Hospital / Bridge / School
#      with criticality used as tie-breaker
#  14. Ranked candidates use the same corrected selection logic
#  15. Recommendation section remains untouched
#  16. Full backup before modification
#  17. Python compilation required before writing
# =============================================================================

import os
import re
import ast
from datetime import datetime

APP_PATH = (
    "/content/gdrive/MyDrive/"
    "CrisisWeave_Project/"
    "05_Streamlit/"
    "CrisisWeave_Live/"
    "live_app.py"
)

print("=" * 78)
print("CRISISWEAVE — SAFE FINAL MULTI-INCIDENT MAP + NEXT VERIFICATION PATCH")
print("=" * 78)

# =============================================================================
# 1. READ CURRENT APP
# =============================================================================

if not os.path.isfile(APP_PATH):
    raise RuntimeError(
        "live_app.py was not found:\n"
        + APP_PATH
    )

with open(
    APP_PATH,
    "r",
    encoding="utf-8"
) as f:
    source = f.read()

print("✓ live_app.py found")
print(
    "  Lines before:",
    len(source.splitlines())
)
print(
    "  Characters  :",
    len(source)
)

# =============================================================================
# 2. BACKUP
# =============================================================================

timestamp = datetime.now().strftime(
    "%Y%m%d_%H%M%S"
)

BACKUP_PATH = (
    APP_PATH
    +
    ".backup_SAFE_FINAL_MAP_"
    +
    timestamp
)

with open(
    BACKUP_PATH,
    "w",
    encoding="utf-8"
) as f:
    f.write(source)

print()
print("✓ Backup created:")
print(
    " ",
    BACKUP_PATH
)

# =============================================================================
# 3. GUARANTEE GLOBAL FOLIUM IMPORTS
# =============================================================================

imports_to_add = []

if "import folium" not in source:
    imports_to_add.append(
        "import folium"
    )

if "from streamlit_folium import st_folium" not in source:
    imports_to_add.append(
        "from streamlit_folium import st_folium"
    )

if imports_to_add:

    import_anchor = (
        "import streamlit as st"
    )

    if import_anchor not in source:
        raise RuntimeError(
            "Could not safely add Folium imports because "
            "'import streamlit as st' was not found.\n"
            "NO CHANGES WERE WRITTEN."
        )

    source = source.replace(
        import_anchor,
        import_anchor
        +
        "\n"
        +
        "\n".join(
            imports_to_add
        ),
        1
    )

    print()
    print(
        "✓ Added global map imports:"
    )

    for item in imports_to_add:
        print(
            "  ",
            item
        )

else:

    print()
    print(
        "✓ Folium imports already present"
    )

# =============================================================================
# 4. LOCATE THE CURRENT MULTI-INCIDENT NEXT-LOCATION BLOCK
# =============================================================================
#
# DO NOT search for "Incident Verification Map".
#
# That heading has changed through the previous patches.
#
# The stable structure is:
#
#     priority = incident_result.get(
#
# followed later by:
#
#     # RECOMMENDATION
#
# We replace ONLY that region.
# =============================================================================

priority_anchor = (
    "priority = incident_result.get("
)

priority_positions = [
    match.start()
    for match in re.finditer(
        re.escape(priority_anchor),
        source
    )
]

if not priority_positions:

    raise RuntimeError(
        "\nCould not find the multi-incident "
        "'priority = incident_result.get(...)' block.\n"
        "NO CHANGES WERE WRITTEN."
    )

priority_start = priority_positions[-1]

recommendation_marker = (
    "# RECOMMENDATION"
)

recommendation_position = source.find(
    recommendation_marker,
    priority_start
)

if recommendation_position == -1:

    raise RuntimeError(
        "\nCould not find the RECOMMENDATION boundary "
        "after the selected incident priority block.\n"
        "NO CHANGES WERE WRITTEN."
    )

print()
print(
    "✓ Selected-incident priority block found at:",
    priority_start
)

print(
    "✓ Recommendation boundary found at:",
    recommendation_position
)

# =============================================================================
# 5. NEW MAP + NEXT-LOCATION BLOCK
# =============================================================================
#
# IMPORTANT:
# This block is already indented for:
#
#     for incident_id, incident_data in incidents.items():
#
# so it belongs INSIDE the incident loop.
# =============================================================================

new_block = r'''
        # ============================================================
        # NEXT VERIFICATION LOCATION
        # ============================================================

        priority = incident_result.get(
            "priority",
            {}
        )

        # ------------------------------------------------------------
        # USE GIS BELONGING TO THIS INCIDENT
        # ------------------------------------------------------------

        gis_context = incident_data.get(
            "gis_context",
            {}
        )

        gis_features = gis_context.get(
            "features",
            []
        )

        # ------------------------------------------------------------
        # BUILD CORRECT NEXT-VERIFICATION CANDIDATES
        # ------------------------------------------------------------
        #
        # Only:
        #   Hospital
        #   Bridge
        #   School
        #
        # Roads are never candidates.
        #
        # The previous engine ranking could allow water proximity and
        # coarse distance buckets to make a farther school win over a
        # nearer school.
        #
        # Here:
        #
        #   proximity = PRIMARY
        #   criticality = SECONDARY
        #
        # Therefore, for example:
        #
        #   Govt Girls Hr Sec School  ~104 m
        #   Govt Boys Hr Sec School   ~146 m
        #   Govt Elementary School    ~216 m
        #
        # the nearer school is selected instead of the farther one.
        # ------------------------------------------------------------

        criticality_map = {
            "Hospital": 1.00,
            "Bridge": 0.90,
            "School": 0.75,
        }

        candidates = []

        for feature in gis_features:

            feature_type = feature.get(
                "feature_type",
                "Other"
            )

            if feature_type not in {
                "Hospital",
                "Bridge",
                "School"
            }:
                continue

            latitude = feature.get(
                "latitude"
            )

            longitude = feature.get(
                "longitude"
            )

            distance = feature.get(
                "distance_m"
            )

            if (
                latitude is None
                or longitude is None
                or distance is None
            ):
                continue

            try:

                latitude = float(
                    latitude
                )

                longitude = float(
                    longitude
                )

                distance = float(
                    distance
                )

            except (
                TypeError,
                ValueError
            ):

                continue

            criticality = criticality_map.get(
                feature_type,
                0.50
            )

            # Continuous proximity score.
            proximity_score = max(
                0.0,
                1.0 -
                (
                    distance /
                    500.0
                )
            )

            # Proximity is intentionally dominant.
            operational_score = (
                0.75 *
                proximity_score
                +
                0.25 *
                criticality
            )

            candidates.append(
                {
                    "name":
                        str(
                            feature.get(
                                "name",
                                "Unnamed"
                            )
                        ),

                    "feature_type":
                        feature_type,

                    "latitude":
                        latitude,

                    "longitude":
                        longitude,

                    "distance_m":
                        distance,

                    "direction_alignment":
                        feature.get(
                            "direction_alignment",
                            0.50
                        ),

                    "incident_proximity":
                        round(
                            proximity_score,
                            3
                        ),

                    "water_proximity":
                        feature.get(
                            "water_proximity",
                            feature.get(
                                "water_score",
                                0.0
                            )
                        ),

                    "base_criticality":
                        criticality,

                    "priority_score":
                        round(
                            operational_score,
                            3
                        ),
                }
            )

        # ------------------------------------------------------------
        # SORT:
        #
        # 1. Nearer location first
        # 2. Higher criticality if distance is effectively tied
        # 3. Alphabetical final tie-breaker
        # ------------------------------------------------------------

        candidates.sort(
            key=lambda item: (
                item["distance_m"],
                -item["base_criticality"],
                item["name"].lower(),
            )
        )

        selected = (
            candidates[0]
            if candidates
            else None
        )

        # ------------------------------------------------------------
        # UPDATE INCIDENT PRIORITY OBJECT
        # ------------------------------------------------------------

        if selected is not None:

            priority[
                "selected_next_priority"
            ] = selected

            priority[
                "ranked_candidates"
            ] = list(
                candidates
            )

            incident_result[
                "priority"
            ] = priority

        # ============================================================
        # NEXT VERIFICATION LOCATION
        # ============================================================

        st.markdown(
            "### Next Verification Location"
        )

        if selected is not None:

            st.success(
                selected["name"]
                +
                " — "
                +
                selected["feature_type"]
            )

            next_col1, next_col2, next_col3 = (
                st.columns(3)
            )

            with next_col1:

                st.metric(
                    "Distance",
                    f"{selected['distance_m']:.1f} m"
                )

            with next_col2:

                st.metric(
                    "Priority Score",
                    f"{selected['priority_score']:.3f}"
                )

            with next_col3:

                st.metric(
                    "Criticality",
                    f"{selected['base_criticality']:.2f}"
                )

            st.caption(
                "Selected from nearby critical infrastructure "
                "using incident proximity first and "
                "infrastructure criticality as the tie-breaker."
            )

        else:

            st.info(
                "No eligible hospital, bridge, or school "
                "was found in the current GIS context."
            )

        # ============================================================
        # MAP
        # ============================================================
        #
        # No "Incident Verification Map" heading is displayed.
        #
        # RED WARNING
        #     = Incident center
        #
        # PURPLE STAR
        #     = Next verification location
        #
        # GREEN
        #     = School
        #
        # RED PLUS
        #     = Hospital
        #
        # ORANGE BRIDGE ICON
        #     = Bridge
        #
        # BLUE WATER DROP
        #     = Water resource
        #
        # GRAY INFO
        #     = Other useful GIS feature
        #
        # ROADS
        #     = hidden
        # ============================================================

        map_lat = (
            float(center_lat)
            if center_lat is not None
            else float(
                st.session_state.get(
                    "incident_lat",
                    0.0
                )
            )
        )

        map_lon = (
            float(center_lon)
            if center_lon is not None
            else float(
                st.session_state.get(
                    "incident_lon",
                    0.0
                )
            )
        )

        verification_map = folium.Map(
            location=[
                map_lat,
                map_lon
            ],
            zoom_start=15,
            control_scale=True
        )

        # ============================================================
        # 🔴 INCIDENT CENTER
        # ============================================================

        folium.Marker(
            [
                map_lat,
                map_lon
            ],

            tooltip=(
                f"{incident_id} — Incident Center"
            ),

            popup=(
                "<b>🔴 Incident Center</b><br>"
                f"<b>Incident:</b> {incident_id}<br>"
                f"<b>Latitude:</b> {map_lat:.6f}<br>"
                f"<b>Longitude:</b> {map_lon:.6f}"
            ),

            icon=folium.Icon(
                color="red",
                icon="warning-sign"
            )

        ).add_to(
            verification_map
        )

        # ============================================================
        # IDENTIFY SELECTED LOCATION
        # ============================================================

        selected_key = None

        if selected is not None:

            selected_key = (
                selected[
                    "feature_type"
                ],

                round(
                    float(
                        selected[
                            "latitude"
                        ]
                    ),
                    7
                ),

                round(
                    float(
                        selected[
                            "longitude"
                        ]
                    ),
                    7
                )
            )

        # ============================================================
        # GIS MARKERS
        # ============================================================

        for feature in gis_features:

            latitude = feature.get(
                "latitude"
            )

            longitude = feature.get(
                "longitude"
            )

            if (
                latitude is None
                or longitude is None
            ):
                continue

            try:

                latitude = float(
                    latitude
                )

                longitude = float(
                    longitude
                )

            except (
                TypeError,
                ValueError
            ):

                continue

            feature_type = feature.get(
                "feature_type",
                "Other"
            )

            # --------------------------------------------------------
            # ROADS ARE NOT SHOWN
            # --------------------------------------------------------

            if feature_type == "Road":
                continue

            name = str(
                feature.get(
                    "name",
                    "Unnamed"
                )
            )

            distance = feature.get(
                "distance_m"
            )

            current_key = (
                feature_type,
                round(
                    latitude,
                    7
                ),
                round(
                    longitude,
                    7
                )
            )

            # --------------------------------------------------------
            # SELECTED LOCATION MUST APPEAR ONLY AS ⭐
            # --------------------------------------------------------

            if (
                selected_key is not None
                and
                current_key == selected_key
            ):
                continue

            # --------------------------------------------------------
            # 🟢 SCHOOL
            # --------------------------------------------------------

            if feature_type == "School":

                marker_color = "green"
                marker_icon = "education"
                marker_label = "School"

            # --------------------------------------------------------
            # ➕ RED HOSPITAL
            # --------------------------------------------------------

            elif feature_type == "Hospital":

                marker_color = "red"
                marker_icon = "plus"
                marker_label = "Hospital"

            # --------------------------------------------------------
            # 🟠 BRIDGE
            # --------------------------------------------------------

            elif feature_type == "Bridge":

                marker_color = "orange"
                marker_icon = "road"
                marker_label = "Bridge"

            # --------------------------------------------------------
            # 💧 BLUE WATER RESOURCE
            # --------------------------------------------------------

            elif feature_type == "Waterbody":

                marker_color = "blue"
                marker_icon = "tint"
                marker_label = "Water Resource"

            # --------------------------------------------------------
            # ⚪ OTHER GIS FEATURE
            # --------------------------------------------------------

            else:

                marker_color = "gray"
                marker_icon = "info-sign"
                marker_label = "Other GIS Feature"

            popup_text = (
                f"<b>{name}</b><br>"
                f"<b>Type:</b> {marker_label}<br>"
                f"<b>Distance:</b> "
                f"{distance if distance is not None else 'N/A'} m"
            )

            folium.Marker(
                [
                    latitude,
                    longitude
                ],

                tooltip=(
                    f"{marker_label}: {name}"
                ),

                popup=popup_text,

                icon=folium.Icon(
                    color=marker_color,
                    icon=marker_icon
                )

            ).add_to(
                verification_map
            )

        # ============================================================
        # ⭐ NEXT VERIFICATION LOCATION
        # ============================================================

        if selected is not None:

            folium.Marker(
                [
                    selected[
                        "latitude"
                    ],
                    selected[
                        "longitude"
                    ]
                ],

                tooltip=(
                    "⭐ Next Verification — "
                    +
                    selected[
                        "name"
                    ]
                ),

                popup=(
                    "<b>⭐ Next Verification Location</b><br>"
                    f"<b>Name:</b> "
                    f"{selected['name']}<br>"
                    f"<b>Type:</b> "
                    f"{selected['feature_type']}<br>"
                    f"<b>Distance:</b> "
                    f"{selected['distance_m']:.1f} m<br>"
                    f"<b>Priority Score:</b> "
                    f"{selected['priority_score']:.3f}"
                ),

                icon=folium.Icon(
                    color="purple",
                    icon="star"
                )

            ).add_to(
                verification_map
            )

        # ============================================================
        # MAP LEGEND
        # ============================================================

        legend_html = """
        <div style="
            position: fixed;
            bottom: 25px;
            left: 25px;
            z-index: 9999;
            background: white;
            padding: 10px 13px;
            border: 2px solid #777;
            border-radius: 6px;
            font-size: 13px;
            line-height: 1.55;
            box-shadow: 0 1px 5px rgba(0,0,0,0.35);
        ">

            <b>CrisisWeave Verification</b><br>

            <span style="color:#d62728;">
                ●
            </span>
            Incident Center
            <br>

            <span style="color:#9c27b0;">
                ★
            </span>
            Next Verification
            <br>

            <span style="color:#2ca02c;">
                ●
            </span>
            School
            <br>

            <span style="color:#d62728;">
                ✚
            </span>
            Hospital
            <br>

            <span style="color:#ff8c00;">
                ✚
            </span>
            Bridge
            <br>

            <span style="color:#168bd0;">
                💧
            </span>
            Water Resource
            <br>

            <span style="color:#666;">
                ●
            </span>
            Other GIS Feature

        </div>
        """

        verification_map.get_root().html.add_child(
            folium.Element(
                legend_html
            )
        )

        # ============================================================
        # UNIQUE MAP KEY FOR EACH INCIDENT
        # ============================================================
        #
        # THIS IS IMPORTANT.
        #
        # Using:
        #
        #     key="verification_map"
        #
        # for INCIDENT-001 and INCIDENT-002 causes the two maps
        # to share the same Streamlit widget identity.
        #
        # Now:
        #
        #     verification_map_INCIDENT_001
        #     verification_map_INCIDENT_002
        #
        # ============================================================

        safe_incident_key = re.sub(
            r"[^A-Za-z0-9_]+",
            "_",
            str(
                incident_id
            )
        )

        st_folium(
            verification_map,
            width=None,
            height=600,
            key=(
                "verification_map_"
                +
                safe_incident_key
            ),
            returned_objects=[],
            use_container_width=True
        )

        # ============================================================
        # RANKED VERIFICATION CANDIDATES
        # ============================================================

        if candidates:

            with st.expander(
                "View ranked verification candidates",
                expanded=False
            ):

                ranked_rows = []

                for rank, item in enumerate(
                    candidates,
                    start=1
                ):

                    ranked_rows.append(
                        {
                            "Rank":
                                rank,

                            "Location":
                                item[
                                    "name"
                                ],

                            "Type":
                                item[
                                    "feature_type"
                                ],

                            "Distance (m)":
                                round(
                                    item[
                                        "distance_m"
                                    ],
                                    1
                                ),

                            "Priority Score":
                                item[
                                    "priority_score"
                                ],

                            "Criticality":
                                item[
                                    "base_criticality"
                                ],
                        }
                    )

                st.dataframe(
                    pd.DataFrame(
                        ranked_rows
                    ),
                    use_container_width=True,
                    hide_index=True
                )

'''

# =============================================================================
# 6. REPLACE ONLY THE BROKEN NEXT/MAP AREA
# =============================================================================

candidate_source = (
    source[
        :priority_start
    ]
    +
    new_block
    +
    "\n        "
    +
    source[
        recommendation_position:
    ]
)

# =============================================================================
# 7. SAFETY / SYNTAX CHECK
# =============================================================================

print()
print(
    "Running Python syntax check..."
)

try:

    ast.parse(
        candidate_source
    )

except SyntaxError as exc:

    print()
    print(
        "✗ CANDIDATE SOURCE FAILED COMPILATION"
    )

    print(
        f"  Line   : {exc.lineno}"
    )

    print(
        f"  Column : {exc.offset}"
    )

    print(
        f"  Error  : {exc.msg}"
    )

    print()
    print(
        "NO CHANGES WERE WRITTEN."
    )

    raise RuntimeError(
        "Candidate live_app.py did not compile."
    )

print(
    "✓ Candidate Python compilation passed"
)

# =============================================================================
# 8. SAFETY CHECKS
# =============================================================================

required_tokens = {

    "incident marker":
        'icon="warning-sign"',

    "next star":
        'icon="star"',

    "school marker":
        'marker_icon = "education"',

    "hospital marker":
        'marker_icon = "plus"',

    "bridge marker":
        'marker_icon = "road"',

    "water marker":
        'marker_icon = "tint"',

    "gray marker":
        'marker_icon = "info-sign"',

    "unique map key":
        '"verification_map_"',

    "st_folium":
        "st_folium(",

    "candidate selection":
        '"selected_next_priority"',

    "roads excluded":
        'if feature_type == "Road":',

}

print()
print(
    "Running final safety checks..."
)

for label, token in required_tokens.items():

    if token not in candidate_source:

        raise RuntimeError(
            f"Safety check failed: "
            f"{label} token missing: {token}\n"
            "NO CHANGES WERE WRITTEN."
        )

    print(
        "✓",
        label
    )

# =============================================================================
# 9. IMPORTANT CHECK:
#    OLD BROKEN incident_id MAP TEXT MUST NOT REMAIN OUTSIDE LOOP
# =============================================================================
#
# The new map uses incident_id only inside the existing incident loop.
# We do NOT attempt fragile global deletion of every incident_id token.
# =============================================================================

map_count = candidate_source.count(
    "st_folium("
)

if map_count < 2:

    raise RuntimeError(
        "Safety check failed: expected the original incident map "
        "plus the new per-incident verification maps."
    )

print()
print(
    "✓ Existing incident-location map preserved"
)

print(
    "✓ New per-incident verification map added"
)

# =============================================================================
# 10. WRITE ONLY AFTER ALL CHECKS PASS
# =============================================================================

with open(
    APP_PATH,
    "w",
    encoding="utf-8"
) as f:
    f.write(
        candidate_source
    )

print()
print(
    "✓ live_app.py written successfully"
)

# =============================================================================
# 11. FINAL COMPILE OF ACTUAL FILE
# =============================================================================

with open(
    APP_PATH,
    "r",
    encoding="utf-8"
) as f:
    final_source = f.read()

try:

    ast.parse(
        final_source
    )

except SyntaxError as exc:

    # Restore backup immediately.
    with open(
        BACKUP_PATH,
        "r",
        encoding="utf-8"
    ) as f:
        backup_source = f.read()

    with open(
        APP_PATH,
        "w",
        encoding="utf-8"
    ) as f:
        f.write(
            backup_source
        )

    print()
    print(
        "✗ Final file compilation failed."
    )

    print(
        "✓ Original file automatically restored from backup."
    )

    raise RuntimeError(
        "Final live_app.py compilation failed; "
        "backup restored."
    )

# =============================================================================
# 12. FINAL REPORT
# =============================================================================

print()
print("=" * 78)
print("PATCH COMPLETE — SAFE")
print("=" * 78)

print(
    "✓ Final Python compilation passed"
)

print(
    "✓ Final lines:",
    len(
        final_source.splitlines()
    )
)

print(
    "✓ Final characters:",
    len(
        final_source
    )
)

print()
print(
    "MAP BEHAVIOR:"
)

print(
    "  🔴 Incident center"
)

print(
    "  ⭐ Next verification location"
)

print(
    "  🟢 School"
)

print(
    "  ➕ Red hospital"
)

print(
    "  🟠 Bridge"
)

print(
    "  💧 Blue water resource"
)

print(
    "  ⚪ Other GIS feature"
)

print(
    "  🚫 Roads hidden"
)

print()
print(
    "NEXT LOCATION:"
)

print(
    "  ✓ Nearest Hospital / Bridge / School is selected"
)

print(
    "  ✓ Criticality is used as secondary tie-breaker"
)

print(
    "  ✓ Selected location appears only as purple ⭐"
)

print(
    "  ✓ Each incident has its own unique map key"
)

print()
print(
    "BACKUP:"
)

print(
    BACKUP_PATH
)

print()
print("=" * 78)
print("DO NOT MODIFY live_app.py AGAIN BEFORE TESTING THIS VERSION")
print("=" * 78)
```
